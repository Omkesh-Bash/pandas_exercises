### Step 1. Import the necessary libraries

```python
import pandas as pd
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/justmarkham/DAT8/master/data/u.user). 

### Step 3. Assign it to a variable called users.

```python
users = pd.read_csv('https://raw.githubusercontent.com/justmarkham/DAT8/master/data/u.user', sep = '|', index_col= 'user_id')
users.head()
```

### Step 4. Discover what is the mean age per occupation

```python
users.groupby('occupation').age.mean()
```

### Step 5. Discover the Male ratio per occupation and sort it from the most to the least

```python
users['numeric_gender'] = users['gender'].apply(lambda x : 1 if x =='M' else 0)
df = users.groupby('occupation').numeric_gender.sum() / users.occupation.value_counts() * 100
df.sort_values(ascending= False)
```

### Step 6. For each occupation, calculate the minimum and maximum ages

```python
# users.groupby('occupation').agg(
#     min_age = ('age', 'min'),
#     max_age = ('age', 'max')
# )

users.groupby('occupation').age.agg(['min', 'max'])
```

### Step 7. For each combination of occupation and gender, calculate the mean age

```python
users.groupby(['occupation', 'gender']).age.mean().reset_index()
```

### Step 8.  For each occupation present the percentage of women and men

```python
df = users.groupby('occupation').numeric_gender.sum() / users.occupation.value_counts() * 100
df = df.reset_index().rename(columns= {0 : 'male_percentage'})
df['female_percentage'] = 100 - df['male_percentage']
df
```

```python
gender_occu = users.groupby(['occupation', 'gender']).gender.agg('count')
# gender_occu
occupation_count = users.occupation.value_counts(sort = False)
# occupation_count
percentage_gender_occu = gender_occu.div(occupation_count, level = 'occupation') * 100
percentage_gender_occu.reset_index()
```
