### Step 1. Import the necessary libraries

```python
import pandas as pd
import numpy as np
```

### Step 2. Import the dataset from this [address](https://archive.ics.uci.edu/ml/machine-learning-databases/wine/wine.data). 

### Step 3. Assign it to a variable called wine

```python
wine = pd.read_csv('https://archive.ics.uci.edu/ml/machine-learning-databases/wine/wine.data')
wine.head()
```

### Step 4. Delete the first, fourth, seventh, nineth, eleventh, thirteenth and fourteenth columns

```python
# type(wine.columns[0])
# wine.columns[[0,3,6,8,11,12,13]]
```

```python
wine = wine.drop(wine.columns[[0,3,6,8,11,12,13]], axis = 1)
wine
```

### Step 5. Assign the columns as below:

The attributes are (donated by Riccardo Leardi, riclea '@' anchem.unige.it):  
1) alcohol  
2) malic_acid  
3) alcalinity_of_ash  
4) magnesium  
5) flavanoids  
6) proanthocyanins  
7) hue 

```python
wine.columns = ['alcohol', 'malic_acid', 'alcalinity_of_ash', 'magnesium', 'flavanoids', 'proanthocyanins', 'hue']
```

```python
wine
```

### Step 6. Set the values of the first 3 rows from alcohol as NaN

```python
wine.iloc[:3, 0] = np.nan
wine.head()
```

### Step 7. Now set the value of the rows 3 and 4 of magnesium as NaN

```python
wine.loc[[2, 3], 'magnesium'] = np.nan
wine.head()
```

### Step 8. Fill the value of NaN with the number 10 in alcohol and 100 in magnesium

```python
# wine.alcohol.fillna(10, inplace = True)
# wine.magnesium.fillna(100, inplace = True)
# wine.head()
```

```python
wine['alcohol'] = wine['alcohol'].fillna(10)
wine['magnesium'] = wine['magnesium'].fillna(100)
wine
```

### Step 9. Count the number of missing values

```python
wine.isnull().sum()
```

### Step 10.  Create an array of 10 random numbers up until 10

```python
num_arr = np.random.randint(low = 0, high = 10, size = 10, dtype = 'l')
num_arr
```

### Step 11.  Use random numbers you generated as an index and assign NaN value to each of cell.

```python
wine.loc[num_arr, 'alcohol'] = np.nan
wine.head()
```

### Step 12.  How many missing values do we have?

```python
wine.isnull().sum()
```

### Step 13. Delete the rows that contain missing values

```python
wine.dropna(inplace= True)
wine.isnull().sum()
```

### Step 14. Print only the non-null values in alcohol

```python
mask = wine.alcohol.notnull()
```

```python
wine.alcohol[mask]
```

### Step 15.  Reset the index, so it starts with 0 again

```python
wine.reset_index(inplace=True, drop= True)
wine
```
